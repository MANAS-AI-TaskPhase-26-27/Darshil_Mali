```python
def predict(X, w, b):
    return X @ w + b
```

```python
def compute_cost(X, y, w, b):
    m = len(y)
    
    predictions = predict(X, w, b)
    errors = predictions - y
    
    cost = (1 / (2 * m)) * np.sum(errors ** 2)
    
    return cost
```

```python
cost = compute_cost(X_train, y_train, w, b)
cost
```

```python
def compute_gradients(X, y, w, b):
    m = len(y)

    predictions = predict(X, w, b)
    errors = predictions - y

    dw = (1 / m) * (X.T @ errors)
    db = (1 / m) * np.sum(errors)

    return dw, db
```

```python
def gradient_descent(X, y, w, b, learning_rate, epochs):
    cost_history = []

    for epoch in range(epochs):

        dw, db = compute_gradients(X, y, w, b)

        w = w - learning_rate * dw
        b = b - learning_rate * db

        cost = compute_cost(X, y, w, b)
        cost_history.append(cost)

    return w, b, cost_history
```

```python
def mean_squared_error(y, predictions):
    m = len(y)
    errors = predictions - y
    return np.sum(errors ** 2) / m
```

```python
def root_mean_squared_error(y, predictions):
    mse = mean_squared_error(y, predictions)
    return np.sqrt(mse)
```

```python
def r_squared(y, predictions):
    residual_sum = np.sum((y - predictions) ** 2)
    total_sum = np.sum((y - np.mean(y)) ** 2)

    return 1 - (residual_sum / total_sum)
```

```python
train_mse = mean_squared_error(y_train, train_predictions)
train_rmse = root_mean_squared_error(y_train, train_predictions)
train_r2 = r_squared(y_train, train_predictions)

val_mse = mean_squared_error(y_val, val_predictions)
val_rmse = root_mean_squared_error(y_val, val_predictions)
val_r2 = r_squared(y_val, val_predictions)
```

```python
print("Training MSE:", train_mse)
print("Training RMSE:", train_rmse)
print("Training R²:", train_r2)

print("\nValidation MSE:", val_mse)
print("Validation RMSE:", val_rmse)
print("Validation R²:", val_r2)
```

```python
def gradient_descent(X_train, y_train, X_val, y_val, w, b, learning_rate, epochs):
    cost_history = []
    train_r2_history = []
    val_r2_history = []

    for epoch in range(epochs):

        # 1. Calculate gradients
        dw, db = compute_gradients(X_train, y_train, w, b)

        # 2. Update weights and bias
        w = w - learning_rate * dw
        b = b - learning_rate * db

        # 3. Make predictions
        train_predictions = predict(X_train, w, b)
        val_predictions = predict(X_val, w, b)

        # 4. Calculate training cost
        cost = compute_cost(X_train, y_train, w, b)

        # 5. Calculate R²
        train_r2 = r_squared(y_train, train_predictions)
        val_r2 = r_squared(y_val, val_predictions)

        # 6. Store everything
        cost_history.append(cost)
        train_r2_history.append(train_r2)
        val_r2_history.append(val_r2)

    return w, b, cost_history, train_r2_history, val_r2_history
```
